```python
import logging
import os
import sys
from pickle import Unpickler
import numpy as np

log = logging.getLogger(__name__)

def loadTrainExamples(name, path):
    examplesFile= path + '/' + name + '.examples'
    if not os.path.isfile(examplesFile):
        log.warning(f'File "{examplesFile}" with trainExamples not found!')
    else:
        log.info("File with trainExamples found. Loading it...")
        with open(examplesFile, "rb") as f:
            trainExamplesHistory = Unpickler(f).load()
        log.info('Loading done!')
        return trainExamplesHistory
    
def get_best_move_position(pi, board_x=19, board_y=19):
    """
    傳入動作機率分布 pi，回傳最大機率對應的棋盤位置 (x, y)，或 'pass'
    """
    best_action = np.argmax(pi)
    if best_action == board_x * board_y:
        return "pass"
    x = best_action // board_y
    y = best_action % board_y
    return (x, y)


def get_top_k_move_positions(pi, board_x=19, board_y=19, k=5):
    """
    傳入動作機率分布 pi，回傳機率前 k 高的動作位置與機率文字描述。
    """
    # 找出前 k 大的索引（從大到小排序）
    topk_indices = np.argsort(pi)[-k:][::-1]
    result = ""
    for idx in topk_indices:
        prob = pi[idx]
        if prob == 0:
            break
        if idx == board_x * board_y:
            result+=f"pass with probability {prob:.4f}\n"
        else:
            x = idx // board_y
            y = idx % board_y
            result+=f"({x}, {y}) with probability {prob:.4f}\n"

    return result

def print_board_pretty(board):
    """
    將棋盤 (numpy array) 轉成人類可讀格式並印出。
    1 -> X（先手），-1 -> O（後手），0 -> 空格。
    """
    size = board.shape[0]
    print("    " + " ".join(f"{i:2}" for i in range(size)))
    print("   " + "---" * size)
    for r in range(size):
        row_str = f"{r:2} |"
        for c in range(size):
            val = board[r][c]
            symbol = {1: "X", -1: "O", 0: "."}.get(val, "?")
            row_str += f" {symbol} "
        print(row_str)
    print()
def print_board_pretty2(board):
    """
    將棋盤 (numpy array) 轉成人類可讀格式並印出。
    1 -> X（先手），-1 -> O（後手），0 -> 空格。
    """
    size = board.shape[0]
    print("    " + " ".join(f"{i:2}" for i in range(size)))
    print("   " + "---" * size)
    for r in range(size):
        row_str = f"{r:2} |"
        for c in range(size):
            val = board[r][c]
            symbol = {1: "O", -1: "X", 0: "."}.get(val, "?")
            row_str += f" {symbol} "
        print(row_str)
    print()
import numpy as np
```
